Tuples:

تاپل‌ها دنباله‌های مرتب و غیرقابل تغییر (immutable) از عناصر هستند. شبیه لیست‌ها ساخته می‌شوند اما به جای کروشه [] از پرانتز () استفاده می‌شود. چون غیرقابل تغییرند، بعد از ساخت نمی‌توان عناصرشان را اضافه، حذف یا ویرایش کرد. برای مواقعی که نمی‌خواهیم داده‌ها تصادفاً تغییر کنند، مناسب هستند.

# my_tuple[0] = 5  # این خط باعث TypeError می‌شود


Dictionaries:

```python
دیکشنری‌ها مجموعه‌های نامرتب (در نسخه‌های جدیدتر پایتون مرتب هستند) از زوج‌های کلید-مقدار (key-value pairs) هستند. با آکولاد {} تعریف می‌شوند و هر کلید باید منحصر به فرد باشد. برای ذخیره اطلاعات با نام مشخص (مثل اطلاعات یک فرد) بسیار مفیدند. به جای ایندکس عددی، با کلیدها به مقادیر دسترسی پیدا می‌کنیم.
# ساخت یک دیکشنری
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(my_dict)

print(my_dict['name'])
print(my_dict['age'])

my_dict['email'] = 'alice@example.com'
my_dict['age'] = 31  # تغییر مقدار موجود
print(my_dict)
```

Sets

```python
مجموعه‌ها (ست‌ها) دسته‌های مرتب نشده‌ای از عناصر منحصر به فرد هستند. با آکولاد {} ساخته می‌شوند، اما اگر خالی باشند باید با set() ساخته شوند تا با دیکشنری اشتباه گرفته نشوند. کاربرد اصلی آن‌ها در حذف مقادیر تکراری و انجام عملیات ریاضی مجموعه‌ها (اشتراک، اجتماع، تفاضل) است.
# ساخت یک ست (با حذف تکراری‌ها)
my_set = {1, 2, 2, 3, 4, 4, 4}
print(my_set) # خروجی: {1, 2, 3, 4} (مرتب نشده)

empty_set = set()
print(empty_set)

my_set.add(5)
print(my_set)

set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1.union(set2))      # اجتماع: {1, 2, 3, 4, 5}
print(set1.intersection(set2)) # اشتراک: {3}
```

## Dictionaries

<div style="direction:rtl">
در دیکشنری های زیر با استفاده از کلید ها، واژه ی 'Hello' را فراخوانی کنید

```python
# Grab 'hello'
d = {'simple_key':'hello'}

# فراخوانی مقدارِ 'hello' با استفاده از کلید
print(d['simple_key'])

```

```python
d = {'k1':{'k2':'hello'}}

# دسترسی به 'hello'
print(d['k1']['k2'])

```

```python
d = {'k1':[{'nest_key':['this is deep',['hello']]}]}

# مسیر دسترسی:
# d['k1']  → لیست
# d['k1'][0]  → دیکشنری
# d['k1'][0]['nest_key']  → لیست ['this is deep', ['hello']]
# d['k1'][0]['nest_key'][1]  → لیست ['hello']
# د عنصر 0 این لیست 'hello' است
print(d['k1'][0]['nest_key'][1][0])

```

```python
d = {'k1':[1,2,{'k2':['this is tricky',{'tough':[1,2,['hello']]}]}]}

# مسیر دسترسی:
# d['k1']               → لیست
# d['k1'][2]            → {'k2': [...]}
# d['k1'][2]['k2']      → لیست ['this is tricky', {'tough': [...]}]
# d['k1'][2]['k2'][1]   → {'tough': [1, 2, ['hello']]}
# d['k1'][2]['k2'][1]['tough']    → لیست [1, 2, ['hello']]
# d['k1'][2]['k2'][1]['tough'][2] → ['hello']
# عنصر 0 این لیست:
print(d['k1'][2]['k2'][1]['tough'][2][0])

```

## List Comprehension, Zip, Enumerate

 <div style="direction:rtl">
از range() استفاده کنید تا تمام اعداد زوج کوچکتر از 10 را نمایش دهید

```python
# نمایش اعداد زوج کوچکتر از 10
for i in range(0, 10, 2):
    print(i)
```

<div style="direction:rtl">
از List comprehension استفاده کنید تا تمام اعداد کوچکتر از 50 که به 3 بخش پذیرند را نمایش دهید

```python
# لیست اعداد کوچکتر از 50 که بر 3 بخش‌پذیرند
multiples_of_3 = [i for i in range(50) if i % 3 == 0]
print(multiples_of_3)
```

<div style="direction:rtl">
با استفاده از zip دو لیست زیر را با هم در یک لیست دیگر ترکیب کنید

```python
l1 = ['a','b','c','d','e','f']
l2 = list('ABCDEF')

# ترکیب دو لیست با zip، تبدیل به لیست از تاپل‌ها
combined = list(zip(l1, l2))
print(combined)
```

<div style="direction:rtl">
از enumerate استفاده کرده و ایندکس هر المان لیست زیر را با خودش ترکیب کنید

```python
l3 = ['a','b','c','d','e','f']

# استفاده از enumerate برای ساخت لیست (index, value)
indexed_l3 = list(enumerate(l3))
print(indexed_l3)
```

```python
l3 = ['a','b','c','d','e','f']

for idx, val in enumerate(l3):
    print(idx, val)      # یا print(f"{idx}: {val}")
```

<div style="direction:rtl">
برنامه ای بنویسید که اعداد صحیح را از 1 تا 100 چاپ کنید. اما برای مضارب سه به جای خود عدد،  "Fizz" را چاپ کند و برای مضارب پنج چاپ به جای خود عدد،  "Buzz" را چاپ کند و برای اعدادی که مضرب هر دو هستند به جای خود عدد،  "FizzBuzz" را چاپ کند.

```python
for num in range(1, 101):
    if num % 3 == 0 and num % 5 == 0:
        print("FizzBuzz")
    elif num % 3 == 0:
        print("Fizz")
    elif num % 5 == 0:
        print("Buzz")
    else:
        print(num)
```

<div style="direction:rtl">
    برنامه ای بنویسید که شتخیص دهد یک عدد اول است یا خیر

```python
# دریافت عدد و تشخیص اول بودن آن
n = int(input("Enter a number: "))

if n < 2:
    print("Not Prime")
else:
    is_prime = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not Prime")
```

<div style="direction:rtl">
برنامه ای بنویسید که اعداد اول بین 2 تا 100 را چاپ کند
راهنمایی : از حلقه های تو در تو استفاده کنید

```python
# چاپ اعداد اول بین 2 تا 100 با حلقه های تو در تو
for num in range(2, 101):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

<div style="direction:rtl">
برای ایجاد لیستی از اولین حرفهای هر کلمه در رشته زیر ، از یک  List comprehension استفاده کنید:

```python
st = 'Create a list of the first letters of every word in this string'
```

```python
# ایجاد لیستی از اولین حروف هر کلمه در رشته با List comprehension
first_letters = [word[0] for word in st.split()]
print(first_letters)
```
